Rainfall Prediction using Logistic Regression

```python
#import libraries
import pandas as pd 
from sklearn.linear_model import LogisticRegression
```

```python
#create the dataset
data={
    "Temperature":[30,32,28,25,24,35,27,29],
    "Humidity":[70,65,80,90,85,60,88,75],
    "Rain":[1,0,1,1,1,0,1,0]             #1:Rain , 0:No Rain

}
df= pd.DataFrame(data)


#define input x and output y 
X=df[["Temperature", "Humidity"]]
y=df["Rain"]
```

```python
#create the logistic regression model
model=LogisticRegression()

#train the model
model.fit(X,y)

#predict  rainfall for new weather condition
new_data=[[26,82]]    #TEMP=26, HUMIDITY= 82%

prediction=model.predict(new_data)
```

```python
#display the result 
print("New Weather Condition:")
print("Temperature =",new_data[0][0],"c")
print("Humidity =",new_data[0][1],"%")


if prediction[0] == 1:
    print("\nPrediction: Rain is Expected")
else:
    print("\nPrediction: No Rain")
```
