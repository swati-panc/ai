```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = {"Age": [19, 21, 20, 23, 31, 35, 40, 42, 45, 48,
            25, 27, 29, 32, 36, 38, 22, 24, 50, 52],
"Annual_Income": [15, 16, 17, 18, 40, 42, 45, 48, 50, 52,
                      25, 27, 30, 32, 35, 38, 20, 22, 55, 58],
"Spending_Score": [90, 85, 88, 80, 60, 55, 50, 45, 40, 35,
                       75, 70, 65, 60, 55, 50, 82, 78, 30, 25]}
df = pd.DataFrame(data)
print("Customer Dataset:")
print(df)
X = df[["Annual_Income", "Spending_Score"]]
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10)
df["Cluster"] = kmeans.fit_predict(X)
print("\nClustered Customer Data:")
print(df)
print("\nCluster Centers:")
print(kmeans.cluster_centers_)
plt.scatter(
    df["Annual_Income"],
    df["Spending_Score"],
    c=df["Cluster"])
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker="X",
    s=200)
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using K-Means")
plt.show()
```
